# Tutorial for Using Split-Raster for GIS (GeoTIFF) Images

This demo focuses on splitting large GeoTIFF (GIS) images into small tiles while preserving geospatial metadata.

### Setup Environment

We recommend using `uv` to manage your environment. For GIS support, you can install the package with the `geo` extra:

```bash
uv sync --extra geo
```

Or using pip:
```bash
pip install "splitraster[geo]"
```

On macOS, you can install GDAL via Homebrew if needed for system libraries:
```bash
brew install gdal
```

````python
import shutil
import warnings
from pathlib import Path

# Suppress GDAL future warnings for cleaner output
warnings.filterwarnings("ignore", category=FutureWarning)

BASE_DIR = Path().resolve().parent
DATA_DIR = BASE_DIR / "tests" / "data"
# print(f"Base Directory: {BASE_DIR}")
# print(f"Data Directory: {DATA_DIR}")
````

````python
from splitraster import geo

# Clean the output folder
shutil.rmtree(DATA_DIR / "processed" / "RGB_TIF", ignore_errors=True)
shutil.rmtree(DATA_DIR / "processed" / "GT_TIF", ignore_errors=True)

input_image_path = str(DATA_DIR / "raw" / "TIF" / "RGB5k.tif")
gt_image_path = str(DATA_DIR / "raw" / "TIF" / "GT5k.tif")

save_path = str(DATA_DIR / "processed" / "RGB_TIF")
save_path_gt = str(DATA_DIR / "processed" / "GT_TIF")

crop_size = 256
repetition_rate = 0
overwrite = True

n = geo.split_image(
    input_image_path,
    save_path,
    crop_size,
    repetition_rate=repetition_rate,
    overwrite=overwrite,
)
print(f"{n} tiles sample of RGB TIF are added at {save_path}")

n = geo.split_image(
    gt_image_path,
    save_path_gt,
    crop_size,
    repetition_rate=repetition_rate,
    overwrite=overwrite,
)
print(f"{n} tiles sample of GT TIF are added at {save_path_gt}")
````

````python
files = sorted([f.name for f in (DATA_DIR / "processed" / "RGB_TIF").iterdir() if f.is_file()])
print(f"Generated {len(files)} files. First 10: {files[:10]}")
````

## Random Sampling with GIS Data

````python
shutil.rmtree(DATA_DIR / "processed" / "Rand" / "RGB_TIF", ignore_errors=True)
shutil.rmtree(DATA_DIR / "processed" / "Rand" / "GT_TIF", ignore_errors=True)

input_save_path = str(DATA_DIR / "processed" / "Rand" / "RGB_TIF")
gt_save_path = str(DATA_DIR / "processed" / "Rand" / "GT_TIF")

n = geo.random_crop_image(
    input_image_path,
    input_save_path,
    gt_image_path,
    gt_save_path,
    crop_size=500,
    crop_number=20,
    img_ext=".tif",
    label_ext=".tif",
    overwrite=True,
)

print(f"{n} sample pairs added at {input_save_path} and {gt_save_path}.")
````
